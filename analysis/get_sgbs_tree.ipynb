```python
from Bio import Phylo
import pandas as pd
import bz2
```

```python
abundances = pd.read_table('/shares/CIBIO-Storage/CM/scratch/users/aitor.blancomiguez/analyses/nw_metanalysis/nw_profiles.tsv')
abundances.index = abundances['Unnamed: 0']
abundances = abundances.drop('Unnamed: 0',1)
keep = list()
for x in abundances.index.tolist():
    keep.append('t__' in x)
sgb_abundances = abundances.loc[keep]
sgb_abundances = sgb_abundances.transpose()
sgb_abundances = sgb_abundances.drop(sgb_abundances.loc[sgb_abundances.sum(axis=1) == 0].index.tolist())
sgb_abundances
```

```python
ab_thresh = 0.001
keep = list()
for x in abundances.index.tolist():
    keep.append('t__' in x)
sgb_prevalences = abundances.loc[keep]
sgb_prevalences = sgb_prevalences.drop(sgb_prevalences.loc[sgb_prevalences.sum(axis=1) == 0].index.tolist())
sgb_prevalences[sgb_prevalences >= ab_thresh] = 1
sgb_prevalences[sgb_prevalences < ab_thresh] = 0
sgb_prevalences
```

```python
to_keep = list()
for sgb in sgb_prevalences.index:
    to_keep.append(sgb.split('|')[-1].replace('t__SGB','').replace('_group',''))
len(to_keep)
```

```python
tree = Phylo.read("/shares/CIBIO-Storage/CM/scratch/users/aitor.blancomiguez/analyses/nw_metanalysis/utils/mpa_v4beta_Jan21_SGB_tree.nwk", "newick")
```

```python
to_remove = list()
for terminal in tree.get_terminals():
    if terminal.name not in to_keep:
        to_remove.append(terminal.name)
print(len(to_remove))
print(len(tree.get_terminals()))
```

```python
for terminal in to_remove:
    tree.prune(target=terminal)
print(len(tree.get_terminals()))
```

```python
Phylo.write(tree, "/shares/CIBIO-Storage/CM/scratch/users/aitor.blancomiguez/analyses/nw_metanalysis/utils/full_sgbs.newick", "newick")
```

```python
fisher = pd.read_csv('/shares/CIBIO-Storage/CM/scratch/users/aitor.blancomiguez/analyses/nw_metanalysis/nw_fisher_test.tsv', sep='\t')
fisher.index = [x.split('|')[-1].replace('t__SGB','').replace('_group','') for x in fisher['Unnamed: 0'].tolist()]
fisher = fisher.drop('Unnamed: 0',1)
fisher
```

```python
phylum = dict()
phylum2count = dict()
sgb2unk = dict()
with bz2.open('/shares/CIBIO-Storage/CM/scratch/databases/MetaRefSGB/releases/Jun23/SGB.Jun23.txt.bz2', 'rt') as read_file:
    for line in read_file:
        if line.startswith('SGB'):
            line = line.strip().split('\t')
            sgb2unk[line[1]]= line[7]
            phylum[line[1]] = line[9].split('|')[1].replace('p__','')
            if line[9].split('|')[1].replace('p__','') not in phylum2count:
                phylum2count[line[9].split('|')[1].replace('p__','')] = 0
            phylum2count[line[9].split('|')[1].replace('p__','')] += 1
        elif line.startswith('GGB'):
            break
phylum2count
```

```python
phylum2color = dict()
phylum2color['Firmicutes'] = '#b69fc9'
phylum2color['Euryarchaeota'] = '#afa9cf'
phylum2color['Actinobacteria'] = '#3c63a4'
phylum2color['Bacteroidota'] = '#528f28'
phylum2color['Proteobacteria'] = '#af1e21'
phylum2color['Spirochaetes'] = '#d27063'
phylum2color['Verrucomicrobia'] = '#fbffaa'
phylum2color['Tenericutes'] = '#e8b05b'
phylum2color['Fusobacteria'] = '#cbe6b9'
phylum2color['Candidatus_Melainabacteria'] = '#884a25'
phylum2color['Candidatus_Saccharibacteria'] = '#502984'
phylum2color['Synergistetes'] = '#a6c0d9'
phylum2color['Elusimicrobia'] = '#b3d960'
phylum2color['Candidatus_Thermoplasmatota'] = '#cccccc'
```

```python
tree = Phylo.read('/shares/CIBIO-Storage/CM/scratch/users/aitor.blancomiguez/analyses/nw_metanalysis/utils/all_sgbs.newick', "newick")
with open('/shares/CIBIO-Storage/CM/scratch/users/aitor.blancomiguez/analyses/nw_metanalysis/utils/all_graphlan_metadata.tsv', 'w') as write_file:
    for terminal in tree.get_terminals():
        write_file.write(terminal.name+"\tclade_marker_edge_width\t0.2\n")
        if terminal.name in fisher.index.tolist():
            if fisher.loc[terminal.name]['direction'] == False:
                write_file.write(terminal.name+'\tring_color\t1\t#f1632a\n')
            elif fisher.loc[terminal.name]['direction'] == True:
                write_file.write(terminal.name+'\tring_color\t1\t#004b79\n')
        if phylum[terminal.name] in phylum2color:
            color = phylum2color[phylum[terminal.name]]
        else:
            color = '#000000'
        write_file.write(terminal.name+"\tclade_marker_color\t"+color+"\n")
```
